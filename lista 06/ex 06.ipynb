## Classe Stack

```python
class Stack:
    def __init__(self):
        self.__itens = []

    def push(self, item):
        self.__itens.append(item)

    def pop(self):
        if self.is_empty():
            raise IndexError("pop em pilha vazia")
        return self.__itens.pop()

    def peek(self):
        if self.is_empty():
            return None
        return self.__itens[-1]

    def is_empty(self):
        return len(self.__itens) == 0

    def size(self):
        return len(self.__itens)

    def __str__(self):
        return str(self.__itens)
```

## Exercicio 5

```python
def decimal_para_hexa(n):
    if n < 0:
        raise ValueError("n deve ser não-negativo")
    if n == 0:
        return "0"
    digitos = "0123456789ABCDEF"
    pilha = Stack()
    while n > 0:
        pilha.push(digitos[n % 16])   # empilha os restos
        n //= 16
    resultado = ""
    while not pilha.is_empty():
        resultado += pilha.pop()      # desempilha na ordem inversa
    return resultado

print(decimal_para_hexa(26))    # 1A
print(decimal_para_hexa(255))   # FF
print(decimal_para_hexa(4096))  # 1000
print(decimal_para_hexa(0))     # 0
print(decimal_para_hexa(26) == hex(26)[2:].upper())  # conferência: True
```

## Exercicio 6

```python
def balanceia_colchetes(expressao):
    pares = {")": "(", "}": "{", "]": "["}
    pilha = Stack()
    for c in expressao:
        if c in "({[":
            pilha.push(c)
        elif c in ")}]":
            if pilha.is_empty() or pilha.pop() != pares[c]:
                return False
    return pilha.is_empty()

testes = ["({[]})", "({[}])", "(a + b) * {c - [d / e]}", "((a + b)", "a + b)", ""]
for t in testes:
    print(f"{t!r:30} -> {'balanceada' if balanceia_colchetes(t) else 'NÃO balanceada'}")
```
