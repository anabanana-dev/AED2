## Exercício 4: Classe Livro

```python
class Livro:
    def __init__(self, titulo, autor, paginas):
        self.titulo = titulo
        self.autor = autor
        self.paginas = paginas
        
    def descricao(self):
        return f"{self.titulo} - {self.autor} ({self.paginas} páginas)"

# Teste
livro1 = Livro("A biblioteca da meia-noite", "Matt Haig", 250)
livro2 = Livro("O Cão dos Baskervilles ", " Arthur Conan Doyle", 300)

print(livro1.descricao())
print(livro2.descricao())
```

## Exercício 5: Conta simples com encapsulamento

```python
class Conta:
    def __init__(self, titular):
        self.titular = titular
        self._saldo = 0  
        
    def depositar(self, valor):
        if valor > 0:
            self._saldo += valor
        else:
            print("O valor de depósito deve ser positivo.")
            
    def consultar_saldo(self):
        return self._saldo

# Teste
conta = Conta("Ana")
conta.depositar(150.00)
conta.depositar(50.50)
print(f"Saldo atual: R$ {conta.consultar_saldo():.2f}")
```

## Exercício 6: Animais - herança e polimorfismo

```python
class Animal:
    def __init__(self, nome):
        self.nome = nome
        
    def emitir_som(self):
        pass

class Cachorro(Animal):
    def emitir_som(self):
        return "au au"

class Gato(Animal):
    def emitir_som(self):
        return "miau"

# Teste com polimorfismo
animais = [Cachorro("Rex"), Gato("Mingau")]

for animal in animais:
    print(f"O {animal.nome} faz: {animal.emitir_som()}")
```

## Exercício 7: Veículo com velocidade controlada

```python
class Veiculo:
    def __init__(self, placa, velocidade_max):
        self.__placa = placa
        self.__velocidade_max = velocidade_max
        self.__velocidade_atual = 0

    @property
    def placa(self):
        return self.__placa

    @property
    def velocidade_max(self):
        return self.__velocidade_max

    @velocidade_max.setter
    def velocidade_max(self, nova_maxima):
        if nova_maxima > 0 and nova_maxima >= self.__velocidade_atual:
            self.__velocidade_max = nova_maxima
        else:
            print("A nova velocidade máxima é inválida.")

    def acelerar(self, incremento=10):
        if self.__velocidade_atual + incremento <= self.__velocidade_max:
            self.__velocidade_atual += incremento
        else:
            self.__velocidade_atual = self.__velocidade_max

    def frear(self, decremento=10):
        if self.__velocidade_atual - decremento >= 0:
            self.__velocidade_atual -= decremento
        else:
            self.__velocidade_atual = 0

    def __str__(self):
        return f"Placa: {self.__placa} | Vel. Atual: {self.__velocidade_atual}km/h | Vel. Máx: {self.__velocidade_max}km/h"

# Teste
carro = Veiculo("ABC-1234", 120)
print(carro)
carro.acelerar(50)
print(carro)
carro.acelerar(100)
print(carro)
carro.frear(30)
print(carro)
```

## Exercício 8: Funcionários e cálculo de bônus

```python
class Funcionario:
    def __init__(self, nome, salario):
        self.nome = nome
        self.salario = salario
        
    def calcular_bonus(self):
        return self.salario * 0.05
        
    def __str__(self):
        return f"Nome: {self.nome} | Salário: R$ {self.salario:.2f}"

class Vendedor(Funcionario):
    def __init__(self, nome, salario, vendas):
        super().__init__(nome, salario)
        self.vendas = vendas
        
    def calcular_bonus(self):
        return (self.salario * 0.05) + (self.vendas * 0.02)

class Gerente(Funcionario):
    def __init__(self, nome, salario, equipe):
        super().__init__(nome, salario)
        self.equipe = equipe
        
    def calcular_bonus(self):
        return (self.salario * 0.10) + (self.equipe * 100)

# Teste com lista de funcionários (polimorfismo)
equipe = [
    Funcionario("Ana", 3000),
    Vendedor("Carlos", 2000, 50000),
    Gerente("Beatriz", 8000, 5)
]

for func in equipe:
    print(f"{func.nome} ({type(func).__name__}) - Bônus: R$ {func.calcular_bonus():.2f}")
```

## Exercício 9: Produtos e contador de objetos

```python
class Produto:
    contador = 0

    def __init__(self, nome, preco):
        Produto.contador += 1
        self.__id = Produto.contador
        self.__nome = nome
        self.__preco = preco

    @property
    def id(self):
        return self.__id

    @property
    def nome(self):
        return self.__nome

    @property
    def preco(self):
        return self.__preco

    def set_preco(self, novo_preco):
        if novo_preco > 0:
            self.__preco = novo_preco
        else:
            print("O preço deve ser maior que zero.")

    def aplicar_desconto(self, percentual):
        if 0 <= percentual <= 100:
            desconto = self.__preco * (percentual / 100)
            self.__preco -= desconto
        else:
            print("O percentual de desconto é inválido.")

    def __str__(self):
        return f"[{self.__id}] {self.__nome} - R$ {self.__preco:.2f}"

# Teste
p1 = Produto("Monitor", 800.00)
p2 = Produto("Teclado Mecânico", 250.00)
p3 = Produto("Mouse Sem Fio", 120.00)

p1.aplicar_desconto(10)
print(p1)
print(p2)
print(p3)
print(f"Total de objetos criados: {Produto.contador}")
```

## Exercício 10: Mini sistema de pedidos de e-commerce

```python
class Produto:
    contador = 0
    def __init__(self, nome, preco):
        Produto.contador += 1
        self.__id = Produto.contador
        self.__nome = nome
        self.__preco = preco

    @property
    def id(self):
        return self.__id

    @property
    def nome(self):
        return self.__nome

    @property
    def preco(self):
        return self.__preco

    @preco.setter
    def preco(self, novo_preco):
        if novo_preco > 0:
            self.__preco = novo_preco
        else:
            print("Preço inválido.")

    def calcular_frete(self):
        return 0

    def __str__(self):
        return f"[{self.__id}] {self.__nome} (R$ {self.__preco:.2f})"

class ProdutoFisico(Produto):
    def __init__(self, nome, preco, peso):
        super().__init__(nome, preco)
        self.peso = peso
        
    def calcular_frete(self):
        return 5.00 + (2.00 * self.peso)

class ProdutoDigital(Produto):
    def __init__(self, nome, preco, tamanho_mb):
        super().__init__(nome, preco)
        self.tamanho_mb = tamanho_mb

class Cliente:
    def __init__(self, nome, email):
        self.__nome = nome
        self.__email = email

    @property
    def nome(self):
        return self.__nome

    @property
    def email(self):
        return self.__email

    def __str__(self):
        return f"{self.__nome} <{self.__email}>"

class Pedido:
    numero_automatico = 0
    def __init__(self, cliente):
        Pedido.numero_automatico += 1
        self.numero = Pedido.numero_automatico
        self.cliente = cliente
        self.produtos = []
        self.__status = "Aberto"

    def adicionar_produto(self, produto):
        if self.__status == "Aberto":
            self.produtos.append(produto)
        else:
            print(f"ERRO: Não é possível adicionar produtos. O pedido #{self.numero} está {self.__status}.")

    def calcular_total(self):
        return sum(p.preco for p in self.produtos)

    def calcular_frete_total(self):
        return sum(p.calcular_frete() for p in self.produtos)

    def fechar_pedido(self):
        self.__status = "Fechado"

    def __str__(self):
        total_prod = self.calcular_total()
        total_frete = self.calcular_frete_total()
        
        relatorio = f"--- PEDIDO #{self.numero} ---\n"
        relatorio += f"Cliente: {self.cliente}\n"
        relatorio += f"Status: {self.__status}\n"
        relatorio += "Itens do Pedido:\n"
        for p in self.produtos:
            relatorio += f"  - {p} (Frete: R$ {p.calcular_frete():.2f})\n"
        relatorio += "-" * 25 + "\n"
        relatorio += f"Total Produtos: R$ {total_prod:.2f}\n"
        relatorio += f"Total Frete: R$ {total_frete:.2f}\n"
        relatorio += f"TOTAL GERAL: R$ {total_prod + total_frete:.2f}"
        return relatorio

# Teste Integrador
cliente = Cliente("Ana Ribeiro", "ana@teste.com.br")

# Produtos misturados
prod_fisico1 = ProdutoFisico("Livro Lógica de Programação", 80.00, 1.2)
prod_fisico2 = ProdutoFisico("Caderno Universitário", 25.00, 0.8)
prod_digital = ProdutoDigital("Curso de SQL em Databricks (Licença)", 150.00, 200)

# Montando o pedido
meu_pedido = Pedido(cliente)
meu_pedido.adicionar_produto(prod_fisico1)
meu_pedido.adicionar_produto(prod_fisico2)
meu_pedido.adicionar_produto(prod_digital)

# Exibindo e fechando
print(meu_pedido)
print("\n" + "="*30 + "\n")
meu_pedido.fechar_pedido()

# Tentativa de adicionar produto após fechado
prod_digital2 = ProdutoDigital("Pacote de Ícones", 15.00, 10)
meu_pedido.adicionar_produto(prod_digital2)
```
