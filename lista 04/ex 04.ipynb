## Questões teóricas

**1. B**: a busca binária tem melhor caso O(1), quando o alvo está logo no meio, e pior caso O(log n). A busca sequencial tem melhor caso O(1) e pior caso O(n).

**2.** A busca binária compara o alvo com o elemento do meio e descarta metade do vetor. Ela descarta a metade esquerda se o alvo for maior e a metade direita se for menor. Essa decisão só é válida se o vetor estiver ordenado, porque aí existe a garantia de que tudo à esquerda do meio é menor e tudo à direita é maior. Em dados não ordenados essa garantia não existe, e o algoritmo pode descartar justamente a metade onde o alvo está. O resultado seria um **falso negativo**: o algoritmo retorna -1 mesmo com o valor presente. Por exemplo, em [7, 2, 9, 1, 5] buscando 2, o meio é 9. Como 2 < 9, a busca segue para a esquerda ([7, 2]), mas o comportamento depende da sorte e não é confiável.

## Exercicio 3

```python
def busca_binaria(vetor, alvo, esq=0, dir=None):
    if dir is None:
        dir = len(vetor) - 1
    if esq > dir:                 # caso base: segmento vazio
        return -1
    meio = (esq + dir) // 2
    if vetor[meio] == alvo:
        return meio
    elif alvo < vetor[meio]:
        return busca_binaria(vetor, alvo, esq, meio - 1)
    else:
        return busca_binaria(vetor, alvo, meio + 1, dir)

v = [2, 5, 8, 12, 16, 23, 38, 56, 72, 91]
print(busca_binaria(v, 23))   
print(busca_binaria(v, 2))    
print(busca_binaria(v, 91))  
print(busca_binaria(v, 40))   
print(busca_binaria([], 1))   
```

## Exercicio 4

```python
def busca_sequencial_contando(vetor, alvo):
    comparacoes = 0
    for i in range(len(vetor)):
        comparacoes += 1
        if vetor[i] == alvo:
            return i, comparacoes
    return -1, comparacoes

def busca_binaria_contando(vetor, alvo):
    esq, dir = 0, len(vetor) - 1
    comparacoes = 0
    while esq <= dir:
        meio = (esq + dir) // 2
        comparacoes += 1
        if vetor[meio] == alvo:
            return meio, comparacoes
        elif alvo < vetor[meio]:
            dir = meio - 1
        else:
            esq = meio + 1
    return -1, comparacoes

lista = list(range(0, 2000, 2))  
alvo = 1999                       

idx_s, comp_s = busca_sequencial_contando(lista, alvo)
idx_b, comp_b = busca_binaria_contando(lista, alvo)

print(f"Tamanho da lista: {len(lista)} | Alvo: {alvo}")
print(f"Busca sequencial -> resultado: {idx_s} | comparações: {comp_s}")
print(f"Busca binária    -> resultado: {idx_b} | comparações: {comp_b}")
```
