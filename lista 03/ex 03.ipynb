## Exercicio 4

```python
def buscar_cliente(nome, lista_clientes):
    for i in range(len(lista_clientes)):
        if lista_clientes[i] == nome:
            return i
    return -1

clientes = ["João", "Maria", "Pedro", "Mariana", "Paulo"]
for nome in ["Mariana", "Ricardo"]:
    idx = buscar_cliente(nome, clientes)
    if idx != -1:
        print(f"Cliente '{nome}' encontrado no índice {idx}.")
    else:
        print(f"Cliente '{nome}' não encontrado.")
```

## Exercicio 5

```python
def conta_ocorrencias(valor, lista):
    contador = 0
    for item in lista:
        if item == valor:
            contador += 1
    return contador

print(conta_ocorrencias(3, [1, 3, 5, 3, 7]))          # 2
print(conta_ocorrencias(9, [1, 3, 5, 3, 7]))          # 0
print(conta_ocorrencias("a", ["a", "b", "a", "a"]))   # 3
print(conta_ocorrencias(1, []))                       # 0
```

## Exercicio 6

```python
def remove_elemento(lista, valor):
    if not lista:
        return None
    for i in range(len(lista)):
        if lista[i] == valor:
            removido = lista[i]
            # desliza os elementos posteriores uma posição para a esquerda
            for j in range(i, len(lista) - 1):
                lista[j] = lista[j + 1]
            lista.pop()  # remove a última posição, agora duplicada
            return removido
    return None

lista = [4, 8, 15, 16, 23, 8]
print("Antes:", lista)
print("Removido:", remove_elemento(lista, 8))
print("Depois:", lista)                              
print("Remover 99:", remove_elemento(lista, 99))     
print("Lista vazia:", remove_elemento([], 1)) 
```
