## Questões teóricas

**1. C**: `__idade` sofre *name mangling*: o Python renomeia o atributo para `_Pessoa__idade`. Por isso ele não fica acessível diretamente como `obj.__idade` de fora da classe e deve ser acessado por métodos, como getters e setters.

**2. D**: polimorfismo. A mesma chamada `obj.mover()` produz comportamentos diferentes conforme o tipo real do objeto.

**3. B**: a POO agrupa estado (atributos) e comportamento (métodos) em classes e objetos. A programação procedural separa dados e funções.

**4. C** (a incorreta): o Python **suporta herança múltipla**, por exemplo `class C(A, B):`.

## Exercicio 5

```python
class Pessoa:
    def __init__(self, nome, idade):
        self.nome = nome
        self.idade = idade

    def faz_aniversario(self):
        self.idade += 1

    def __str__(self):
        return f"{self.nome}, {self.idade} anos"

p1 = Pessoa("Ana", 20)
p2 = Pessoa("Carlos", 35)

print("Antes:", p1, "|", p2)
p1.faz_aniversario()
p2.faz_aniversario()
print("Depois:", p1, "|", p2)
```

## Exercicio 6

```python
class Lampada:
    def __init__(self):
        self.__estado = False     

    def ligar(self):
        self.__estado = True

    def desligar(self):
        self.__estado = False

    def is_acesa(self):
        return self.__estado

l = Lampada()
print("Acesa?", l.is_acesa())  
l.ligar()
print("Acesa?", l.is_acesa())   
l.desligar()
print("Acesa?", l.is_acesa())  

try:
    print(l.__estado)           
except AttributeError:
    print("Não é possível acessar __estado diretamente.")
```

## Exercicio 7

```python
class Animal:
    def fazer_som(self):
        print("Algum som")

class Cachorro(Animal):
    def fazer_som(self):
        print("Au Au")

class Gato(Animal):
    def fazer_som(self):
        print("Miau")

Animal().fazer_som()
Cachorro().fazer_som()
Gato().fazer_som()
```

## Exercicio 8

```python
def animais_falam(lista_animais):
    for animal in lista_animais:
        animal.fazer_som()      

animais = [Cachorro(), Gato(), Cachorro(), Animal()]
animais_falam(animais)
```
