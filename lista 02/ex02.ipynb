# Resoluções - Capítulo 2: Complexidade de Algoritmos

## Exercícios

**Exercício 1:** **O(1)**. A função acessa apenas o índice 0, levando tempo constante independente do tamanho da lista.

**Exercício 2:** **O(n)**. O loop percorre todos os elementos da lista uma única vez (crescimento linear).

**Exercício 3:** **O(log n)**. A busca binária divide o espaço de busca pela metade a cada iteração.

**Exercício 4:** A **v1** é mais rápida. A `v1` faz dois loops independentes, com complexidade **O(n)**. A `v2` usa `.sort()`, que custa **O(n log n)**.

**Exercício 5:** **O(2^n)**. Cada execução gera duas novas chamadas recursivas, dobrando o processamento a cada nível (crescimento exponencial).

**Exercício 6:** A **primeira abordagem (busca linear)**. Percorrer a lista custa **O(n)**. Ordenar primeiro para depois buscar custaria no mínimo **O(n log n)**.
